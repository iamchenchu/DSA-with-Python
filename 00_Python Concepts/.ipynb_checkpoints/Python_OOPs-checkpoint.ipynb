```python
# Class Definition : Blueprint of an Object and it contains all the methods of the object that we create using the blueprint

class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model 
        self.year = year
        self._milage = 0 # Protected Attribute

    def drive(self, milage):
        self._milage += miles
        return f"Driven {miles} miles"

    def get_info(self):
        return f"{self.year} {self.brand} {self.model}"


class ElectricCar(Car):
    def __init__(self, brand, model, year, battery_size):
        super().__init__(brand, model, year)
        self.battery_size = battery_size

    def charge(self):
        return "Charging the battery..."


#Usage 
my_car = ElectricCar("Tesla", "Model 3", 2024, 75)
print(my_car.get_info())
print(my_car.charge())
        
        
    
```

```python
### Basic Class and Objects

# DEFINING A CLASS
class Dog:
    """This sis a class - the blueprint for all dogs"""
    species = "Canis Familiaris"

    def __init__(self, name, age, breed):
        """Constructor - runs when creating new project"""
        self.name = name
        self.age = age
        self.breed = breed
        self.energy = 100 # Default value

    def bark(self):
        """Method - Something the dog can do here"""
        return f"{self.name} says Woof!"

    def play(self, minutes):
        """Method with parameter"""
        self.energy -= minutes * 2
        return f"{self.name} played for {minutes} minutes. Energy: {self.energy}"


#Creating Objects 
dog1 = Dog("Max", 3, "Golden Retriever")
dog2 = Dog("Balla", 5, "Poodle")
dog3 = Dog("Charlie", 2, "Bulldog")

# Each Object 
print(dog1.name)
print(dog2.name)
print(dog3.name)

    
```

```python
class NeuralLayer:
    """Class representing a layer in a neural network"""
    
    # Class variable
    activation_functions = ['relu', 'sigmoid', 'tanh', 'softmax']
    
    def __init__(self, input_size, output_size, activation='relu'):
        """Each layer object has its own configuration"""
        self.input_size = input_size
        self.output_size = output_size
        self.activation = activation
        
        # Initialize weights for this specific layer
        self.weights = np.random.randn(input_size, output_size) * 0.01
        self.bias = np.zeros((1, output_size))
        
        # Track layer-specific information
        self.forward_cache = None
        self.backward_cache = None
    
    def forward(self, X):
        """Process input through this layer"""
        self.forward_cache = X
        z = np.dot(X, self.weights) + self.bias
        return self._apply_activation(z)
    
    def _apply_activation(self, z):
        """Apply activation function"""
        if self.activation == 'relu':
            return np.maximum(0, z)
        # ... other activations

# Creating different layer objects from the same class
layer1 = NeuralLayer(784, 128, 'relu')      # Input layer
layer2 = NeuralLayer(128, 64, 'relu')       # Hidden layer
layer3 = NeuralLayer(64, 10, 'softmax')     # Output layer

# Each layer object has its own weights and configuration
print(f"Layer 1 shape: {layer1.weights.shape}")  # (784, 128)
print(f"Layer 2 shape: {layer2.weights.shape}")  # (128, 64)
print(f"Layer 3 shape: {layer3.weights.shape}")  # (64, 10)
```
