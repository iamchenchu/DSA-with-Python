##### Concept 1: Variables and Data Types

```python
# Variable is a name that refers to a value in memory 
# In python, types are dynamic so we do not need to clearly define the data type.

x = 10  #int 
y = 3.14 # now float
z = 'hi' # now str

print(x, type(x))
print(y, type(y))
print(z, type(z))
```

### Common Primitive Data Types 

1. `int`   - Integers (Epoch steps, counts, epochs)
2. `float` - Real Numbers (Learning rates, Losses, Metrics)
3. `str`   - Text (file paths, Model Names, Experiment id's)
4. `bool`  - Boolean (True/Fales) for conditions and flags
