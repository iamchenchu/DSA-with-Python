```python
# Design a function that prints index and values of each element
array1 = [9, 7, 8, 5, 3]
def index_function(arr):
    for i in range(len(arr)):
        print(f"Index {i} : {arr[i]}")

index_function(array1)
```

```python
# Create an array of length 'n' where arr[i] = i*i

def array_fun(n): 
    arr = []
    for i in range(n):
        arr.append(i*i)

    return arr


n = 10
result = array_fun(n)
print(result)
```

```python
# Array input, output and update

arr = [31, 31, 21, 8, 28]

#Index based traversing
for i in range(len(arr)):
    print("Index",i,":", arr[i])

```

```python
# direct element traverse 
for x in arr:
    print(x)
```

```python
# Update the Array 

array_update = [1, 3, 10, 11, 13]

array_update.append(10)

print(array_update)

array_update.insert(2, 100)
print(array_update)
```

```python
# Delete element from array
array_delete = [11, 31, 101, 11, 13]

del array_delete[3]

print(array_delete)
```
